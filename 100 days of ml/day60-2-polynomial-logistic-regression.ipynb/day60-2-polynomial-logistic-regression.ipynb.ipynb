```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('ushape.csv')
```

```python
df.head()
```

```python
df.columns = ['x','y','output']
```

```python
df.head()
```

```python
plt.scatter(df['x'],df['y'],c=df['output'])
```

```python
x = df.drop(columns=['output'])
y = df['output'] 
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
lr = LogisticRegression()
lr.fit(x_train,y_train)
```

```python
y_pred1 = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
accuracy_score(y_test,y_pred1)
```

```python
print(classification_report(y_test,y_pred1))
```

```python
pd.DataFrame(confusion_matrix(y_test,y_pred1),columns=['1','0'])
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x.values, y.astype('int').values, lr, legend=2)
```

```python
from sklearn.model_selection import cross_val_score
np.mean(cross_val_score(lr, x.values, y.values, scoring='accuracy', cv=10))
```

```python
from sklearn.preprocessing import PolynomialFeatures
ploy = PolynomialFeatures(degree=3,include_bias=False)
x_train_trans = ploy.fit_transform(x_train)
x_test_trans = ploy.transform(x_test)
```

```python
lr1 = LogisticRegression()
lr1.fit(x_train_trans,y_train)
y_pred2 = lr1.predict(x_test_trans)
```

```python
np.mean(cross_val_score(lr1, x_train_trans, y_train.values, scoring='accuracy', cv=10))
```

```python
def plot_decision_boundary(X,y,degree=1):
    
    poly = PolynomialFeatures(degree=degree)
    X_trf = poly.fit_transform(X)
    
    clf = LogisticRegression()
    clf.fit(X_trf,y)
    
    accuracy = np.mean(cross_val_score(clf,X_trf,y,scoring='accuracy',cv=10))
    
    a=np.arange(start=X[:,0].min()-1, stop=X[:,0].max()+1, step=0.01)
    b=np.arange(start=X[:,1].min()-1, stop=X[:,1].max()+1, step=0.01)


    XX,YY=np.meshgrid(a,b)
    
    input_array=np.array([XX.ravel(),YY.ravel()]).T

    labels=clf.predict(poly.transform(input_array))
    
    plt.contourf(XX,YY,labels.reshape(XX.shape),alpha=0.5)
    plt.scatter(X[:,0],X[:,1], c=y)
    plt.title('Degree = {}, accuracy is {}'.format(degree,np.round(accuracy,4)))
```

```python
plot_decision_boundary(x.values,y.values)
```

```python
plot_decision_boundary(x.values,y.values,degree=2)
```

```python
plot_decision_boundary(x.values,y.values,degree=2)
```
