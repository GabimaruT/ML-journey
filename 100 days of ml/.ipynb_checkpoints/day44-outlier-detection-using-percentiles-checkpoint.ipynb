```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('weight-height.csv')
```

```python
df.head()
```

```python
sns.distplot(df['Height'])
```

```python
sns.boxplot(df['Height'],orient='h')
```

```python
lower = df['Height'].quantile(0.99)
lower
```

```python
upper = df['Height'].quantile(0.01)
upper
```

```python
df[(df['Height'] <= 58.13) | (df['Height']>=74.78)]
```

```python
df_new = df[(df['Height']>= 58.13) & (df['Height'] <= 74.78)]
```

```python
df_new.shape
```

```python
sns.boxplot(df_new['Height'],orient='h')
```

```python
sns.distplot(df_new['Height'])
```

```python
df['Height'] = np.where(
    df['Height'] > upper,
    upper,
    np.where(
        df['Height'] < lower,
        lower,
        df['Height']
    )
)
```

```python
sns.boxplot(df['Height'],orient='h')
```
