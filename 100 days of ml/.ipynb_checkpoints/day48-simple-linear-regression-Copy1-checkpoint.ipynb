```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('placement (1).csv')
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('cgpa')
plt.ylabel('package')
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.iloc[:,0:1],df['package'])
```

```python
x_test
```

```python
y_test
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['cgpa'],df['package'])
plt.plot(x_test,y_pred,color='red')
plt.xlabel('cgpa')
plt.ylabel('package')
```

```python
m = lr.coef_
```

```python
c = lr.intercept_
```

```python
y = m*7.5+c
```
