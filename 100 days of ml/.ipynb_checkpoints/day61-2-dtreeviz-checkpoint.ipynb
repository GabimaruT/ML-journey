```python
clf = DecisionTreeClassifier()
iris = load_iris()

x_train = iris.data
y_train = iris.target
clf.fit(x_train,y_train)
```

```python
from sklearn.tree import plot_tree
plot_tree(clf)
```

```python
import dtreeviz
import warnings
warnings.filterwarnings('ignore')

viz_model = dtreeviz.model(
    clf,
    X_train=x_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    class_names=['setosa', 'versicolor', 'virginica'],
)

viz_model.view(scale=2,fontname="DejaVu Sans")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from dtreeviz.trees import *

df_cars = pd.read_csv("cars.csv")
X, y = df_cars[['wgt']], df_cars['mpg']

dt = DecisionTreeRegressor(max_depth=3, criterion="friedman_mse")
dt.fit(X, y)

fig = plt.figure()
ax = fig.gca()
rtreeviz_univar(dt, X, y, 'wgt', 'mpg', ax=ax)
plt.show()
```

```python
df_cars = pd.read_csv("cars.csv")

df_cars
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
import dtreeviz
import pandas as pd

# Assuming df_cars is already loaded
# df_cars = pd.read_csv('cars.csv')

X = df_cars[['wgt', 'eng']]
y = df_cars['mpg']

# Fit the model
dt = DecisionTreeRegressor(max_depth=3, criterion="friedman_mse")
dt.fit(X, y)

# 1. Create the dtreeviz model wrapper (with custom descriptive feature names)
viz_model = dtreeviz.model(dt,
                           X_train=X,
                           y_train=y,
                           feature_names=['Vehicle Weight', 'Engine Disp'],
                           target_name='mpg')

# 2. Setup your 3D matplotlib axis
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')

# 3. Pass the registered feature names here!
viz_model.rtree_feature_space3D(
    features=['Vehicle Weight', 'Engine Disp'],  # <--- Changed from ['wgt', 'eng']
    ax=ax,
    elev=20,
    azim=25,
    dist=8.2,
    show={'splits', 'title'}
)

plt.show()
```
