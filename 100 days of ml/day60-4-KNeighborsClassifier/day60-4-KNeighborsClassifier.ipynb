```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
```

```python
df
```

```python
df.shape
```

```python
df.drop(columns=['id','Unnamed: 32'],inplace=True)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df.iloc[:,1:],df['diagnosis'],test_size=0.2,random_state=2)
```

```python
x_train.shape
```

```python
col = x_train.columns
```

```python
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()
scale.set_output(transform='pandas')
x_train = scale.fit_transform(x_train)
x_test = scale.transform(x_test)
```

```python
x_train.columns = col
```

```python
x_test.columns = col
```

```python
x_train
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(x_train,y_train)
```

```python
y_pred = knn.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
from sklearn.pipeline import Pipeline
pipe = Pipeline([
    ('stander',StandardScaler()),
    ('model',KNeighborsClassifier())
])
```

```python
pipe.fit(x_train,y_train)
```

```python
score = []
```

```python
for i in range (1,16) :
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(x_train,y_train)
    y_pred = knn.predict(x_test)
    score.append(accuracy_score(y_test,y_pred))
```

```python
score
```

```python
import matplotlib.pyplot as plt
plt.plot(range(1,16),score)
```

```python
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions
from sklearn.linear_model import LogisticRegression # swap with your choice of model

# 1. Select only 2 features for visualization
# Let's pick 'radius_mean' and 'texture_mean'
X_plot = x_train[['radius_mean', 'texture_mean']].values

# 2. Convert string targets ('M' / 'B') into integers (1 / 0)
y_plot = y_train.map({'M': 1, 'B': 0}).values

# 3. Fit the model on just these two features
model = LogisticRegression()
model.fit(X_plot, y_plot)

# 4. Plot the decision boundary
plt.figure(figsize=(10, 6))
plot_decision_regions(X_plot, y_plot, clf=model, legend=2)

plt.xlabel('radius_mean (scaled)')
plt.ylabel('texture_mean (scaled)')
plt.title('Decision Boundary using MLXtend')
plt.show()
```
