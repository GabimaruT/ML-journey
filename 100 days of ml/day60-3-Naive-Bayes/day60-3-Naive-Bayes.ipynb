```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder
from sklearn.naive_bayes import CategoricalNB

# 1. Load data
df = pd.read_csv('play_tennis.csv')

# 2. Separate features and target (dropping the 'day' column)
X = df[['outlook', 'temp', 'humidity', 'wind']]
y = df['play']

# 3. Encode categorical text features into integer categories
feature_encoder = OrdinalEncoder()
X_encoded = feature_encoder.fit_transform(X)

# 4. Encode target ('Yes' -> 1, 'No' -> 0)
target_encoder = LabelEncoder()
y_encoded = target_encoder.fit_transform(y)
```

```python
# Initialize and train the Naive Bayes Classifier
clf = CategoricalNB()
clf.fit(X_encoded, y_encoded)
```

```python
# Define a new query row
new_query = pd.DataFrame([['Sunny', 'Hot', 'High', 'Weak']], 
                         columns=['outlook', 'temp', 'humidity', 'wind'])

# Encode the query using the fitted feature encoder
new_query_encoded = feature_encoder.transform(new_query)

# Predict the class directly (Output will be 0 for 'No' or 1 for 'Yes')
prediction_code = clf.predict(new_query_encoded)
prediction_label = target_encoder.inverse_transform(prediction_code)
print(f"Prediction: {prediction_label[0]}")

# Get the exact mathematical probabilities calculated for [No, Yes]
probabilities = clf.predict_proba(new_query_encoded)
print(f"Probability of 'No': {probabilities[0][0]:.4f}")
print(f"Probability of 'Yes': {probabilities[0][1]:.4f}")
```
