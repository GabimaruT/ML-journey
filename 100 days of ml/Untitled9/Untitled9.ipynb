```python
from sklearn.datasets import *
import graphviz.backend as be
from dtreeviz import *
from IPython.display import Image, display_svg,SVG
from sklearn.tree import DecisionTreeClassifier
```

```python
clf = DecisionTreeClassifier(max_depth=2)
iris = load_iris()

x_train = iris.data
y_train = iris.target
clf.fit(x_train,y_train)
```

```python
from sklearn.tree import plot_tree
plot_tree(clf)
```

```python
# 1. Initialize the dtreeviz adapter model using explicit keyword arguments
viz_model = dtreeviz.model(
    clf,
    X_train=x_train,        # Note: Must use capital 'X_train'
    y_train=y_train,        # Note: Must use lowercase 'y_train'
    feature_names=iris.feature_names,
    class_names=['setosa', 'versicolor', 'virginica']
)

# 2. Render and display the tree visualization
viz_model.view()
```
