```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=3)
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```

```python
class MGDRegressor : 
    def __init__(self,learning_rate=0.1,epoch=1):
        self.coef = None
        self.intercept = None
        self.lr = learning_rate
        self.epoch = epoch

    def fit(self,x_train,y_train):
        self.intercept = 0
        self.coef = np.ones(x_train.shape[1])

        #update all the coef and intercept
        for i in range (self.epoch) :

            y_hat = np.dot(x_train,self.coef) + self.intercept
            intercept_der = -2 * np.mean(y_train - y_hat)
            self.intercept = self.intercept - (self.lr * intercept_der)


            coef_der = -2 * np.dot((y_train - y_hat),x_train)/x_train.shape[0]
            self.coef = self.coef - (self.lr * coef_der)

        print(self.intercept,self.coef)

        

    def predict(self,x_test):
        return np.dot(x_test,self.coef) + self.intercept
    
```

```python
gd = GDRegressor(learning_rate=0.05,epoch=100)
```

```python
gd.fit(x_train,y_train)
```

```python
y_pred = gd.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```
